# Milestone 2: Model Building

**Objective:**  
Develop and train a recommendation model using collaborative filtering.

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
user_item_matrix = pd.read_csv(
    "../Milestone_1/user_item_matrix.csv",
    index_col=0
)

user_item_matrix.head()
```

```python
user_similarity = cosine_similarity(user_item_matrix)

user_similarity_df = pd.DataFrame(
    user_similarity,
    index=user_item_matrix.index,
    columns=user_item_matrix.index
)

user_similarity_df.head()
```

```python
def recommend_products(user_id, top_n=5):
    if user_id not in user_item_matrix.index:
        return "User not found"

    # Step 1: Find similar users
    similar_users = user_similarity_df[user_id].sort_values(ascending=False)[1:6]

    # Step 2: Compute weighted product scores
    weighted_scores = user_item_matrix.loc[similar_users.index].T.dot(similar_users)

    # Step 3: Remove products already purchased by user
    user_purchases = user_item_matrix.loc[user_id]
    weighted_scores = weighted_scores[user_purchases == 0]

    # Step 4: Return top N recommendations
    return weighted_scores.sort_values(ascending=False).head(top_n)
```

```python
sample_user = user_item_matrix.index[0]

print("Recommendations for user:", sample_user)
recommend_products(sample_user)
```
